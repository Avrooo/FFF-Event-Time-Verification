# NAMRC N1 — Four representative raw-history re-audits

**Default does not run simulations.** It searches the extracted N1 folder, checks four representative cases and reruns the existing postprocessor on available histories. Missing raw histories can be regenerated for only these cases by setting `RERUN_MISSING=True`. This is a *repeatability check* of the archived auditor, not an independently coded second energy algorithm.

```
from pathlib import Path
import sys, os, shutil
# Finds the extracted N1 experiment folder even if Jupyter started in your user directory
roots = [Path.home()/"Downloads", Path.cwd()]
found=[]
for r in roots:
    if r.exists():
        found += [p.parent for p in r.rglob("run_N1_complete.py") if (p.parent/"N1_energy_audit.py").exists()]
if not found: raise FileNotFoundError("Extract the N1 package in Downloads")
ROOT=found[0]
SCRIPT=ROOT/"verify_N1_four_cases.py"
print("Experiment folder:",ROOT)
print("Copy verify_N1_four_cases.py from the verification ZIP into this folder before continuing")
assert SCRIPT.exists(), f"Missing {SCRIPT}"
```

```
RERUN_MISSING = False  # Keep False for first run. Set True ONLY if you need missing histories regenerated.
INCLUDE_RAW_IN_ZIP = False  # True makes a much larger review ZIP
import subprocess
args=[sys.executable,"-u",str(SCRIPT)]
if RERUN_MISSING: args.append("--rerun-missing")
if INCLUDE_RAW_IN_ZIP: args.append("--include-raw")
result=subprocess.run(args,cwd=ROOT,check=False)
print("Exit:",result.returncode)
print("Evidence ZIP:", ROOT/"N1_representative_evidence_for_review.zip")
```
