# NAMRC N1 — final full-solver generalization experiment

Runs **36 actual Rust simulations** (two 6-layer toolpaths × 3 speeds × 3 time steps × two timing modes), audits physical-clock energy accounting, and creates a compact results ZIP. The runner retains raw histories for reproducibility; do not run two copies concurrently.

Extract this ZIP first. Rust/Cargo and NumPy are required.

```python
from pathlib import Path
import os,sys
# Locate the extracted experiment directory without assuming Jupyter started in it.
HOME=Path.home()
candidates=[HOME/'Downloads'/'NAMRC_N1_FINAL',HOME/'Downloads'/'NAMRC_N1_Two_Toolpath_Full_Solver_Experiment'/'NAMRC_N1_FINAL',Path.cwd()]
ROOT=next((p for p in candidates if (p/'run_N1_complete.py').exists()),None)
if ROOT is None:
    hits=list((HOME/'Downloads').rglob('run_N1_complete.py'))
    if hits:ROOT=hits[0].parent
if ROOT is None:raise FileNotFoundError('Extract NAMRC_N1_FINAL_One_Click_Windows.zip under Downloads first.')
os.chdir(ROOT)
sys.path.insert(0,str(ROOT))
print('N1 experiment folder:',ROOT)
```

```python
from pathlib import Path
import os,subprocess,sys
os.environ['PATH']=str(Path.home()/'.cargo'/'bin')+os.pathsep+os.environ.get('PATH','')
subprocess.run(['cargo','--version'],check=True)
subprocess.run([sys.executable,'-c','import numpy;print("NumPy OK")'],check=True)
print('Ready to run all 36 cases')
```

```python
import subprocess,sys
# ONE command: compile, run 36 cases, audit all retained histories, write paired results and ZIP.
subprocess.run([sys.executable,'-u',str(ROOT/'run_N1_complete.py')],cwd=ROOT,check=True)
```
